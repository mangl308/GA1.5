# Part 1: Introduction and set up

In this assignment, you will investigate the joint variability of three soil
properties. Soil properties vary naturally from sample to sample due to
differences in soil composition, depositional history, stress history, and
water content. These properties may also be statistically dependent.
Therefore, considering them jointly can provide information that is not
captured when each property is analysed separately.

The dataset **Soil Properties** contains laboratory soil-characterisation
data. In this exercise, we focus on three random variables:

- Unit weight, $\gamma$ ($\mathrm{kN/m^3}$)
- Water content, $w$ (%)
- Plasticity index, $PI$ (%)

```python
import numpy as np
from scipy.stats import multivariate_normal
from scipy.stats import norm
import matplotlib.pyplot as plt
from statistics import mean, stdev
```

```python
import os
from urllib.request import urlretrieve

def findfile(fname):
    if not os.path.isfile(fname):
        print(f'Downloading {fname}...')
        urlretrieve('https://github.com/TUDelft-MUDE/source-files/raw/main/file/'+fname, fname)

fname = 'Soil_Properties.csv'
findfile(fname)
```

> **Task 2.2:**
>
>Quantify the strength of the linear dependence between each pair using Pearson's correlation coefficient. What can you conclude?


```python
def calculate_covariance(X1, X2):
    mean_x1 = np.mean(X1)
    mean_x2 = np.mean(X2)

    # Step 2: calculate deviations from the means
    diff_x1 = X1 - mean_x1
    diff_x2 = X2 - mean_x2

    # Step 3: multiply corresponding deviations
    product = diff_x1*diff_x2

    # Step 4: calculate the sample covariance
    # Hint: because these are sample observations, use N - 1
    covariance = np.sum(product) /(len(X1) - 1)

    return covariance


def pearson_correlation(X1, X2):
    covariance = calculate_covariance(X1, X2)

    # Step 2: calculate the sample standard deviations
    std_x1 = np.std(X1, ddof=1)
    std_x2 = np.std(X2, ddof=1)

    # Step 3: standardize the covariance
    correl_coeff = covariance/(std_x1*std_x2)

    return correl_coeff


print("Pearson's correlation coefficient between γ and w is",
      np.round(pearson_correlation(data[:,0], data[:,1]), 3))

print("Pearson's correlation coefficient between γ and PI is",
      np.round(pearson_correlation(data[:,0], data[:,2]), 3))

print("Pearson's correlation coefficient between w and PI is",
      np.round(pearson_correlation(data[:,1], data[:,2]), 3))
```

**Answer Task 2.1 & 2.2:**

*Marginals (Task 2.1):* $\gamma$ (mean ≈ 15.2 kN/m³, std ≈ 1.6) is roughly symmetric and close to bell-shaped. $w$ (mean ≈ 79 %, std ≈ 42 %) and $PI$ (mean ≈ 38 %, std ≈ 22 %) are clearly **right-skewed** (skewness ≈ 2.5 and 1.4), with a few large values ($w$ up to ≈ 307 %, $PI$ up to ≈ 134 %). Relative to its mean, $\gamma$ varies little (CoV ≈ 0.1), while $w$ and $PI$ vary a lot (CoV ≈ 0.5–0.6).

*Dependence (Task 2.2):*
- $\rho(\gamma, w) \approx -0.85$: strong negative dependence. Wetter soils have more pore water and less solid material, so they are lighter.
- $\rho(\gamma, PI) \approx -0.64$: moderate negative dependence.
- $\rho(w, PI) \approx +0.68$: moderate positive dependence. More plastic (clayey) soils hold more water.

These values match the scatter plots. Pearson's $\rho$ only measures *linear* dependence, and the skewed marginals and outliers can distort it.

## Part 3: Multivariate Gaussian distribution

> **Task 3.1:**
>
>Model the joint distribution of $\gamma$, $w$ and $PI$ using a multivariate Gaussian distribution. Follow the steps:
>- Define the vector of means.
>- Define the covariance matrix.
>- Draw 500 samples from the fitted trivariate Gaussian distribution and compare them with the observations.
>
>Do you see differences between the observations and the Gaussian samples? Compare both the univariate marginal distributions and the bivariate scatter plots.


```python
# ---------------------------------------------------------
# Define the vector of means
# ---------------------------------------------------------

mu1 = np.mean(data[:,0]) #γ
mu2 = np.mean(data[:,1]) #W
mu3 = np.mean(data[:,2]) #PI

mu = np.array([mu1, mu2, mu3])


# ---------------------------------------------------------
# Define the covariance matrix
# ---------------------------------------------------------

s1 = np.std(data[:,0], ddof=1)
s2 = np.std(data[:,1], ddof=1)
s3 = np.std(data[:,2], ddof=1)

covariance_12 = calculate_covariance(data[:,0], data[:,1])
covariance_13 = calculate_covariance(data[:,0], data[:,2])
covariance_23 = calculate_covariance(data[:,1], data[:,2])

sigma = np.array([
    [s1**2, covariance_12, covariance_13],
    [covariance_12, s2**2, covariance_23],
    [covariance_13, covariance_23, s3**2]
])
# ---------------------------------------------------------
# Draw 500 samples from the fitted trivariate Gaussian
# ---------------------------------------------------------

samples = multivariate_normal(mean=mu, cov=sigma).rvs(size=500)

# ---------------------------------------------------------
# Compare samples with observations
# ---------------------------------------------------------

fig, axs = plt.subplots(
    3, 3,
    figsize=(8, 6),
    layout='constrained'
)

for i in range(data.shape[1]):
    for j in range(data.shape[1]):

        # Marginal distributions
        if i == j:

            axs[i, j].grid()

            axs[i, j].hist(
                data[:, i],
                bins=20,
                density=True,
                alpha=0.9,
                edgecolor='black',
                facecolor='none',
                label='Observations'
            )

            axs[i, j].hist(
                samples[:, i],
                bins=20,
                density=True,
                alpha=0.5,
                edgecolor='black',
                label='Gaussian samples'
            )

            axs[i, j].set_xlabel(variable_names[i])
            axs[i, j].set_ylabel('PDF')

            axs[i, j].legend(fontsize=7)


        # Scatter plots
        elif i < j:

            axs[i, j].grid()

            # Gaussian samples first
            axs[i, j].scatter(
                samples[:, i],
                samples[:, j],
                alpha=0.30,
                s=20,
                label='Gaussian samples'
            )

            # Observations on top
            axs[i, j].scatter(
                data[:, i],
                data[:, j],
                alpha=0.9,
                s=30,
                edgecolor='black',
                facecolor='none',
                label='Observations'
            )

            axs[i, j].set_xlabel(variable_names[i])
            axs[i, j].set_ylabel(variable_names[j])

            # Put observations first in legend
            handles, labels = axs[i, j].get_legend_handles_labels()

            axs[i, j].legend(
                [handles[1], handles[0]],
                [labels[1], labels[0]],
                fontsize=7,
                markerscale=0.7
            )

        else:

            axs[i, j].axis('off')

plt.show()
```

**Answer Task 3.1:**

The Gaussian samples reproduce the mean, spread and the direction/strength of linear dependence of the observations quite well (the clouds have the same orientation). Differences:
- **Marginals:** the observed $w$ and $PI$ are right-skewed with a long upper tail, while the Gaussian is symmetric. The model misses the very large observed values and puts too much probability on low values.
- **Physically impossible values:** the Gaussian produces **negative** $w$ and $PI$, which cannot happen.
- **Scatter plots:** the observed $\gamma$–$w$ relation is slightly non-linear (curved, fanning out at low $\gamma$), while the Gaussian can only produce elliptical, linear dependence.

The multivariate Gaussian is a reasonable first approximation for $\gamma$, but less suitable for $w$ and $PI$ (a transformation, e.g. log, or non-Gaussian marginals would be better).

> **Task 3.2:**
>
>Suppose we are interested in soil with the following properties:
>
>- unit weight: $\gamma < 16\ \mathrm{kN/m^3}$
>- water content: $w > 80\%$
>- plasticity index: $PI > 30\%$
>
>Using samples generated from the fitted multivariate Gaussian distribution, estimate the probability that **all three conditions are satisfied at the same time**.
>
>Repeat the calculation using different numbers of samples,
>
>$$
>N = 100,\quad 1{,}000,\quad 10{,}000,\quad 100{,}000,\quad 1{,}000{,}000.
>$$
>
>How does the estimated probability behave as the number of samples increases? Comment on the stability of the estimate.
>
>**Hint:** For each generated sample, check whether all three conditions are satisfied. The probability can be estimated as the proportion of generated samples that satisfy all three conditions.


```python
# Investigate the effect of sample size
sample_sizes = [100, 1000, 10000, 100000, 1000000]

for N in sample_sizes:

    # Draw samples from the fitted multivariate Gaussian distribution
    samples = multivariate_normal(mean=mu, cov=sigma).rvs(size=N)

    # Check the three screening criteria simultaneously
    condition = (samples[:, 0] < 16) & (samples[:, 1] > 80) & (samples[:, 2] > 30)

    # Estimate the probability as the proportion satisfying all three conditions
    prob_meeting_condition = np.sum(condition) / N

    print(f'N = {N:7d}: estimated probability = {prob_meeting_condition:.4f}')
```

**Answer Task 3.2:**

The estimated probability converges to $P \approx 0.41$. For small $N$ (100, 1,000) the estimate fluctuates noticeably from run to run, because the Monte Carlo standard error is $\sqrt{p(1-p)/N}$ (≈ 0.05 for $N=100$, ≈ 0.0005 for $N=10^6$). As $N$ grows the estimate stabilises, and from $N \approx 10^5$ it is stable to about the third decimal. Note that this is the probability *under the fitted Gaussian model*. Model errors (e.g. the skewness in $w$ and $PI$) do not disappear with more samples.

## Part 4: Conditioning the multivariate Gaussian distribution


> **Task 4.1:**
>
>Suppose a laboratory measurement gives a water content of
>
>$$
>w = 120\%
>$$
>
>which is above the sample mean. What does this observation imply for the joint distribution of unit weight $\gamma$ and plasticity index $PI$?
>
>1. Compute the conditional distribution of $\gamma$ and $PI$ given $w = 120\%$.
>
>2. Compute the conditional probability
>
>$$
>P\left(
>\gamma < 16\ \mathrm{kN/m^3},
>\quad
>PI > 30\%
>\mid
>w = 120\%
>\right).
>$$


```python
# Conditional distribution of X = [γ, PI] given Y = w = 120%

a = 120.0

# Partition the mean vector for X = [gamma, PI] and Y = w
mu_X = np.array([
    mu[0],      # mean of gamma
    mu[2]       # mean of PI
])
mu_Y = mu[1]    # mean of w

# Partition the covariance matrix
Sigma_XX = np.array([
    [sigma[0, 0], sigma[0, 2]],
    [sigma[2, 0], sigma[2, 2]]
])

Sigma_XY = np.array([
    [sigma[0, 1]],
    [sigma[2, 1]]
])

Sigma_YY = sigma[1, 1]

# ---------------------------------------------------------
# Conditional Gaussian equations
#
# mu_X|Y = mu_X + Sigma_XY * Sigma_YY^(-1) * (a - mu_Y)
#
# Sigma_X|Y = Sigma_XX
#             - Sigma_XY * Sigma_YY^(-1) * Sigma_YX
# ---------------------------------------------------------

# Conditional mean vector
cond_mu_vector = mu_X + Sigma_XY.flatten() / Sigma_YY * (a - mu_Y)

# Conditional covariance matrix
cond_sigma_matrix = Sigma_XX - Sigma_XY @ Sigma_XY.T / Sigma_YY

# Conditional standard deviations
cond_std = np.sqrt(np.diag(cond_sigma_matrix))

print('Conditional mean [γ, PI] =')
print(np.round(cond_mu_vector, 3))

print('\nConditional covariance matrix =')
print(np.round(cond_sigma_matrix, 3))

print('\nConditional standard deviations [γ, PI] =')
print(np.round(cond_std, 3))

# ---------------------------------------------------------
# Generate samples from the conditional Gaussian distribution
# ---------------------------------------------------------

N = 100000
cond_samples = multivariate_normal(mean=cond_mu_vector, cov=cond_sigma_matrix).rvs(size=N)

# ---------------------------------------------------------
# Estimate P(gamma < 16, PI > 30 | w = 120%)
# ---------------------------------------------------------

condition = (cond_samples[:, 0] < 16) & (cond_samples[:, 1] > 30)
cond_prob = np.sum(condition) / N

print(
    f'\nConditional probability '
    f'P(γ < 16, PI > 30 | w = 120%) = {cond_prob:.4f}'
)
```

**Answer Task 4.1:**

Given $w = 120\%$ (above the mean of ≈ 79 %):
- Conditional mean: $\mu_{\gamma|w} \approx 13.92$ kN/m³ (from 15.24) and $\mu_{PI|w} \approx 52.5\%$ (from 38.1 %).
- Conditional standard deviations: $\sigma_{\gamma|w} \approx 0.85$ (from 1.60) and $\sigma_{PI|w} \approx 15.9$ (from 21.7).
- Conditional covariance $\approx -2.13$, so the conditional correlation is $-2.13/(0.846 \cdot 15.86) \approx -0.16$ (from −0.64).

$P(\gamma < 16, PI > 30 \mid w = 120\%) \approx 0.92$. This is much higher than the unconditional $P(\gamma<16, PI>30) \approx 0.54$, because a high $w$ pushes $\gamma$ down and $PI$ up, which are exactly the directions of the two conditions.

> **Task 4.2:**
>
>Using the results from the previous task, compare the **conditional** and **unconditional** distributions of $\gamma$ and $PI$.
>
>1. Plot the conditional and unconditional **marginal distributions** of $\gamma$ and $PI$.
>
>2. Plot the conditional and unconditional **joint probability density contours** of $\gamma$ and $PI$.

>3. Compare the two cases and explain:

```python
# Compare conditional and unconditional marginals of γ and PI

values_gamma = np.linspace(
    max(0, mu[0] - 4*np.sqrt(sigma[0,0])),
    mu[0] + 4*np.sqrt(sigma[0,0]),
    200
)

uncond_γ_pdf = norm.pdf(values_gamma, loc=mu[0], scale=np.sqrt(sigma[0, 0]))

cond_γ_pdf = norm.pdf(values_gamma, loc=cond_mu_vector[0], scale=cond_std[0])


values_PI = np.linspace(
    max(0, mu[2] - 4*np.sqrt(sigma[2,2])),
    mu[2] + 4*np.sqrt(sigma[2,2]),
    200
)

uncond_PI_pdf = norm.pdf(values_PI, loc=mu[2], scale=np.sqrt(sigma[2, 2]))

cond_PI_pdf = norm.pdf(values_PI, loc=cond_mu_vector[1], scale=cond_std[1])


# Plot marginal distributions
fig, ax = plt.subplots(1, 2, figsize=(9, 4), layout='constrained')

# γ
ax[0].plot(
    values_gamma,
    uncond_γ_pdf,
    label='Unconditional distribution'
)

ax[0].plot(
    values_gamma,
    cond_γ_pdf,
    '--',
    label='Conditional distribution'
)

ax[0].grid()
ax[0].set_xlabel(r'$\gamma$ (kN/m$^3$)')
ax[0].set_ylabel('PDF')
ax[0].legend()


# PI
ax[1].plot(
    values_PI,
    uncond_PI_pdf,
    label='Unconditional distribution'
)

ax[1].plot(
    values_PI,
    cond_PI_pdf,
    '--',
    label='Conditional distribution'
)

ax[1].grid()
ax[1].set_xlabel(r'$PI$ (%)')
ax[1].set_ylabel('PDF')
ax[1].legend()

plt.show()
```

```python
# Compare joint conditional and unconditional distributions of γ and PI

# ---------------------------------------------------------
# 1. Create values for γ and PI
# ---------------------------------------------------------

values_gamma = np.linspace(9, 21, 200)
values_PI = np.linspace(0, 120, 200)

# Create a two-dimensional grid of γ and PI
G, I = np.meshgrid(values_gamma, values_PI)


# ---------------------------------------------------------
# 2. Define the unconditional distribution of [γ, PI]
# ---------------------------------------------------------

uncond_mu = [
    mu[0],      # mean of γ
    mu[2]       # mean of PI
]

uncond_cov = np.array([
    [sigma[0, 0], sigma[0, 2]],
    [sigma[2, 0], sigma[2, 2]]
])


# ---------------------------------------------------------
# 3. Calculate the joint PDFs
# ---------------------------------------------------------

# Create pairs of [γ, PI] values
X = np.column_stack([np.concatenate(G.T), np.concatenate(I.T)])
# Hint: use np.concatenate on G.T and I.T, then stack the two 1D arrays as columns.

# Unconditional joint PDF
Z_uncond = multivariate_normal(mean=uncond_mu, cov=uncond_cov).pdf(X)

# Conditional joint PDF given w = 120%
Z_cond = multivariate_normal(mean=cond_mu_vector, cov=cond_sigma_matrix).pdf(X)

# Convert the PDF values back to the two-dimensional grid
Z_uncond = Z_uncond.reshape(G.T.shape).T
Z_cond = Z_cond.reshape(G.T.shape).T


# ---------------------------------------------------------
# 4. Plot the contours
# ---------------------------------------------------------

fig, ax = plt.subplots(figsize=(7, 5))

ax.contour(
    G, I, Z_uncond,
    levels=8,
    linestyles='-'
)

ax.contour(
    G, I, Z_cond,
    levels=8,
    linestyles='--'
)

ax.grid()

ax.set_xlabel(r'$\gamma$ (kN/m$^3$)')
ax.set_ylabel(r'$PI$ (%)')

ax.set_title(
    r'Unconditional and conditional distributions ($w=120\%$)'
)

plt.show()
```

**Answer Task 4.2:**

- **Means:** $\gamma$ shifts *down* (15.2 → 13.9 kN/m³) because $\rho(\gamma,w)<0$, and $PI$ shifts *up* (38 → 52 %) because $\rho(w,PI)>0$. The shift is $\Sigma_{XY}\Sigma_{YY}^{-1}(a-\mu_Y)$, so it is proportional to the covariance with $w$ and to how far $w=120\%$ lies from its mean.
- **Uncertainty:** both variances decrease, and the conditional variance does not depend on the observed value $a$. The variance is reduced by a factor $(1-\rho^2)$: for $\gamma$, $1-0.85^2 \approx 0.28$ (std 1.60 → 0.85, a large reduction); for $PI$, $1-0.68^2 \approx 0.53$ (std 21.7 → 15.9, a smaller reduction). The stronger the correlation with $w$, the more information knowing $w$ gives.
- **Joint shape:** the conditional contours are smaller, shifted to lower $\gamma$/higher $PI$, and much less tilted. Most of the dependence between $\gamma$ and $PI$ came *through* $w$, so once $w$ is known their remaining correlation is only ≈ −0.16.

# Summary of answers (test notes)

## Data and key formulas
- 127 observations of $\gamma$ (kN/m³), $w$ (%), $PI$ (%).
- Sample covariance: $\text{Cov}(X_1,X_2) = \frac{1}{N-1}\sum (x_{1,i}-\bar x_1)(x_{2,i}-\bar x_2)$, using $N-1$ because these are sample data.
- Pearson correlation: $\rho = \dfrac{\text{Cov}(X_1,X_2)}{\sigma_1\sigma_2}$, with $-1 \le \rho \le 1$. It measures **linear** dependence only.
- Conditional Gaussian, with $X = [\gamma, PI]$ and $Y = w = a$:
  - $\mu_{X|Y} = \mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(a - \mu_Y)$
  - $\Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}$
  - For one variable: $\sigma^2_{\gamma|w} = \sigma^2_\gamma(1-\rho^2_{\gamma w})$ and $\mu_{\gamma|w} = \mu_\gamma + \rho\frac{\sigma_\gamma}{\sigma_w}(a-\mu_w)$

## Part 3.1: Multivariate Gaussian fit
- Parameters: mean vector $\mu$ = sample means; covariance matrix $\Sigma$ with variances on the diagonal and covariances off the diagonal (symmetric).
- The fit captures the means, spread and **direction/strength of the linear dependence** well.
- The fit fails because:
  - it cannot produce the **skewness and long upper tails** of $w$ and $PI$;
  - it gives **negative $w$ and $PI$**, which are physically impossible;
  - it only allows **linear, elliptical** dependence, while the $\gamma$–$w$ relation is slightly curved.
- So it is OK for $\gamma$ but poor for $w$ and $PI$. Better options: a log-transform or non-Gaussian marginals.

## Part 3.2: Monte Carlo joint probability
- $P(\gamma<16,\ w>80,\ PI>30) \approx 0.41$. Estimate it as the number of samples satisfying **all** conditions divided by $N$.
- Small $N$ gives noisy estimates. The standard error is $\sqrt{p(1-p)/N}$: ≈ 0.05 at $N=100$ and ≈ 0.0005 at $N=10^6$. The estimate is stable from $N \approx 10^5$.
- More samples reduce the **sampling error, not the model error**.

## Part 4.2: Interpretation
